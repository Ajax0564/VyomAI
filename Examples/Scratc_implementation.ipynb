## 📌 Model Functions

### 1. Prediction Function

The prediction is computed as:

$$
\hat{y} = X \cdot m + b
$$

- $X$: Input feature matrix  
- $m$: Weight vector  
- $b$: Bias term  
- $\hat{y}$: Predicted output  

---

### 2. Cost Function

#### Base Cost (Mean Squared Error)

$$
J = \frac{1}{n} \sum ( \hat{y} - y )^2
$$

- $n$: Number of samples  
- $y$: True values  
- $\hat{y}$: Predicted values  

---

#### With L2 Regularization (Ridge)

$$
J = \frac{1}{n} \sum ( \hat{y} - y )^2 + \frac{\lambda}{2n} \sum m^2
$$

---

#### With L1 Regularization (Lasso)

$$
J = \frac{1}{n} \sum ( \hat{y} - y )^2 + \frac{\lambda}{2n} \sum |m|
$$

- $\lambda$: Regularization parameter (`self.reg`)  
- $m$: Model weights  

## 📌 Training (Gradient Descent)

### Initialization

- Weights initialization:
$$
m = \mathbf{1}
$$

- Bias initialization:
$$
b = 0
$$

- Number of samples:
$$
n = \text{number of rows in } y
$$

---

$$
J = \frac{1}{n} \sum ( \hat{y} - y )^2
$$

### Gradient Descent Update Rule

For each iteration:

#### 1. Prediction

$$
\hat{y} = X \cdot m + b
$$

---

#### 2. Gradient w.r.t Weights

**Without Regularization:**

$$
\frac{\partial J}{\partial m} = \frac{1}{n} X^T (\hat{y} - y)
$$

---

**With L2 Regularization (Ridge):**

$$
\frac{\partial J}{\partial m} = \frac{1}{n} X^T (\hat{y} - y) + \frac{\lambda}{n} m
$$

---

**With L1 Regularization (Lasso):**

$$
\frac{\partial J}{\partial m} = \frac{1}{n} X^T (\hat{y} - y) + \frac{\lambda}{n} \, \text{sign}(m)
$$

---

#### 3. Gradient w.r.t Bias

$$
\frac{\partial J}{\partial b} = \frac{1}{n} \sum (\hat{y} - y)
$$

---

#### 4. Parameter Updates

- Weights update:
$$
m = m - \alpha \cdot \frac{\partial J}{\partial m}
$$

- Bias update:
$$
b = b - \alpha \cdot \frac{\partial J}{\partial b}
$$

---

### Notes

- $\alpha$: Learning rate (`learning_rate`)  
- $\lambda$: Regularization parameter (`self.reg`)  
- $\text{sign}(m)$:
  - $-1$ if $m < 0$
  - $0$ if $m = 0$
  - $1$ if $m > 0$

---

### Cost Tracking

At each iteration, the cost is computed as:

$$
J = \frac{1}{n} \sum ( \hat{y} - y )^2 \; (+ \text{regularization term if applied})
$$

The cost is stored and printed every 100 iterations.

````python
class SimpleLinearRegression:
    def __init__(self, regtype=None):

        self.reg = 1  # l2-norm
        self.regtype = regtype

    def predict(self, X):
        return np.dot(X, self.m) + self.b

    def get_cost(self, y_pred, y_true):
        cost = (np.sum(np.square(y_pred - y_true))) / (self.size)

        if self.regtype == "l2":
            return cost + np.sum((self.reg * np.sum(self.m**2))) / (2 * self.size)

        if self.regtype == "l1":
            return cost + np.sum((self.reg * np.abs(self.m))) / (2 * self.size)
        return cost

    def fit(self, X, y, learning_rate=1e-3, iterations=2500):

        self.size = y.shape[0]
        self.b = 0

        cost = []
        self.m = np.ones((X.shape[1], 1))
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0) + 1e-8
        X = (X - self.mean) / self.std

        for i in range(iterations):
            y_pred = self.predict(X)
            cost.append(self.get_cost(y_pred, y))
            dm = np.dot(X.T, (y_pred - y)) / self.size

            if self.regtype == "l2":
                dm += (self.reg * self.m) / self.size

            if self.regtype == "l1":
                dm += (self.reg * np.sign(self.m)) / self.size


            db = np.sum(y_pred - y) / self.size

            self.m = self.m - (learning_rate * dm)
            self.b = self.b - (learning_rate * db)

            if (i + 1) % 100 == 0:
                print("Cost at iteration number " + str(i + 1) + " is : ", cost[i])
        plt.title("cost v/s iteration")
        plt.plot(cost)
        plt.xlabel("no. of iterations")
        plt.ylabel("cost")
        plt.show()
````

````python
model = SimpleLinearRegression(regtype="l2")
model.fit(x, y)
````

**Logistic Regression**

## 📌 Gradient Derivation (Logistic Regression) aka Binary classification

This section shows how we derive the gradients used in training:
- $dm = \frac{\partial J}{\partial m}$
- $db = \frac{\partial J}{\partial b}$

---

### 1. Model Definition

#### Linear Step
$$
z = X \cdot m + b
$$

#### Sigmoid Activation
$$
\hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
$$

---

### 2. Cost Function (Binary Cross-Entropy)

$$
J = -\frac{1}{n} \sum \left[ y \log(\hat{y}) + (1 - y)\log(1 - \hat{y}) \right]
$$

---

### 3. Chain Rule Breakdown

We compute gradients using the chain rule:

$$
\frac{\partial J}{\partial m} = \frac{\partial J}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z} \cdot \frac{\partial z}{\partial m}
$$

$$
\frac{\partial J}{\partial b} = \frac{\partial J}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z} \cdot \frac{\partial z}{\partial b}
$$

---

### 4. Step-by-Step Derivatives

#### (a) Derivative of Cost w.r.t $\hat{y}$

$$
\frac{\partial J}{\partial \hat{y}} = -\left( \frac{y}{\hat{y}} - \frac{1 - y}{1 - \hat{y}} \right)
$$

---

#### (b) Derivative of Sigmoid

A key identity:

$$
\frac{d}{dz} \sigma(z) = \sigma(z)(1 - \sigma(z))
$$

So,

$$
\frac{\partial \hat{y}}{\partial z} = \hat{y}(1 - \hat{y})
$$

---

#### (c) Derivative of Linear Term

$$
\frac{\partial z}{\partial m} = X
$$

$$
\frac{\partial z}{\partial b} = 1
$$

---

### 5. Combine Terms

Now multiply all parts:

#### For Weights

$$
\frac{\partial J}{\partial m}
= \left[-\left( \frac{y}{\hat{y}} - \frac{1 - y}{1 - \hat{y}} \right)\right]
\cdot \hat{y}(1 - \hat{y}) \cdot X
$$

This simplifies neatly to:

$$
\frac{\partial J}{\partial m} = X^T (\hat{y} - y) / n
$$

---

#### For Bias

$$
\frac{\partial J}{\partial b}
= \frac{1}{n} \sum (\hat{y} - y)
$$

---

### 6. Final Gradient Expressions (Used in Code)

$$
dm = \frac{1}{n} X^T (\hat{y} - y)
$$

$$
db = \frac{1}{n} \sum (\hat{y} - y)
$$

---

### 7. With Regularization

#### L2 (Ridge)

$$
dm = \frac{1}{n} X^T (\hat{y} - y) + \frac{\lambda}{n} m
$$

---

#### L1 (Lasso)

$$
dm = \frac{1}{n} X^T (\hat{y} - y) + \frac{\lambda}{n} \, \text{sign}(m)
$$

---

### 🔑 Key Insight

The combination of:
- sigmoid derivative: $\hat{y}(1 - \hat{y})$
- cross-entropy loss

**cancels out nicely**, giving a very simple gradient:  

$$
\hat{y} - y
$$

logistic regression gradients look similar to linear regression, even though the model is non-linear.

````python
class SimpleLogisticRegression:
    def __init__(self, regtype=None):
        self.reg = 1  # l2-norm
        self.regtype = regtype

    def predict(self, X):
        z = np.dot(X, self.m) + self.b
        y_pred = self._sigmoid(z)
        eps = 1e-8
        y_pred = np.clip(y_pred, eps, 1 - eps)
        return y_pred

    def _sigmoid(self, z):
        z = np.clip(z, -500, 500)
        return 1 / (1 + np.exp(-z))

    def get_cost(self, y_pred, y_true):
        cost = (-1 / self.size) * np.sum(
            y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
        )

        if self.regtype == "l2":
            return cost + (np.sum((self.reg * np.sum(self.m**2))) / (2 * self.size))

        if self.regtype == "l1":
            return cost + (np.sum((self.reg * np.abs(self.m))) / (2 * self.size))
        return cost

    def fit(self, X, y, learning_rate=1e-3, iterations=100):
        self.size = X.shape[0]
        self.b = 0

        cost = []
        self.m = np.ones((X.shape[1], 1))
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0) + 1e-8
        X = (X - self.mean) / self.std

        for i in range(iterations):
            y_pred = self.predict(X)
            # print(y.shape, y_pred.shape)
            cost.append(self.get_cost(y_pred, y))
            dm = np.dot(X.T, (y_pred - y)) / self.size
            db = np.sum(y_pred - y) / self.size
            
            if self.regtype == "l1":
                dm += (self.reg * np.sign(self.m)) / self.size
            if self.regtype == "l2":
                dm += (self.reg * self.m) / self.size

            self.m = self.m - (learning_rate * dm)
            self.b = self.b - (learning_rate * db)

            if (i + 1) % 25 == 0:
                print("Cost at iteration number " + str(i + 1) + " is : ", cost[i])
        plt.title("cost v/s iteration")
        plt.plot(cost)
        plt.xlabel("no. of iterations")
        plt.ylabel("cost")
        plt.show()
````

````python
x, y = make_classification(random_state=1, n_features=5, n_classes=2, n_samples=1000)
y = y.reshape(-1, 1)
print(x.shape, y.shape)
````

````python
model = SimpleLogisticRegression(regtype="l2")
model.fit(x, y)
````

````python
x_norm = (x - model.mean) / model.std
y_pred = model.predict(x_norm)
y_pred_labels = (y_pred >= 0.5).astype(int)

cm = confusion_matrix(y, y_pred_labels)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.show()
````

## 📌 Softmax (Multiclass Logistic Regression)

---

### 1. Model Definition

#### Linear Transformation

$$
Z = X \cdot M + b
$$

- $X \in \mathbb{R}^{n \times d}$: Input data  
- $M \in \mathbb{R}^{d \times k}$: Weight matrix  
- $b \in \mathbb{R}^{k}$: Bias vector  
- $Z \in \mathbb{R}^{n \times k}$: Class scores  

---

### 2. Softmax Function

To convert scores into probabilities:

#### Numerically Stable Softmax

$$
Z_{\text{shift}} = Z - \max(Z)
$$

$$
\text{softmax}(Z_i) = \frac{e^{Z_i}}{\sum_j e^{Z_j}}
$$

Full expression:

$$
\hat{y}_{i,j} = \frac{e^{Z_{i,j} - \max(Z_i)}}{\sum_{k} e^{Z_{i,k} - \max(Z_i)}}
$$

- $\hat{y}_{i,j}$: Probability of class $j$ for sample $i$

---

### 3. Cost Function (Cross-Entropy Loss)

For class labels $y$:

$$
J = \frac{1}{n} \sum_{i=1}^{n} -\log(\hat{y}_{i, y_i})
$$

---

#### With L2 Regularization

$$
J = \frac{1}{n} \sum_{i=1}^{n} -\log(\hat{y}_{i, y_i}) + \frac{\lambda}{2} \sum M^2
$$

- $\lambda$: Regularization parameter (`alpha` in code)

---

### 4. Gradient Derivation

#### One-Hot Encoding Insight

Let $Y$ be the one-hot encoded labels.

---

### 5. Key Result (After Derivation)

The gradient simplifies to:

$$
\delta = \hat{Y} - Y
$$

Where:
- $\hat{Y}$: Predicted probabilities (softmax output)  
- $Y$: True labels (one-hot encoded)

---

### 6. Gradients

#### Gradient w.r.t Weights

$$
\frac{\partial J}{\partial M} = X^T (\hat{Y} - Y)
$$

---

#### Gradient w.r.t Bias

$$
\frac{\partial J}{\partial b} = \sum (\hat{Y} - Y)
$$

---

### 7. With L2 Regularization

$$
\frac{\partial J}{\partial M} = X^T (\hat{Y} - Y) + \lambda M
$$

---

### 8. Parameter Updates

- Weights:
$$
M = M - \eta \cdot \frac{\partial J}{\partial M}
$$

- Bias:
$$
b = b - \eta \cdot \frac{\partial J}{\partial b}
$$

- $\eta$: Learning rate (`learning_rate`)

---

### 9. Implementation Insight

Instead of explicitly computing one-hot encoding:

$$
\delta = \hat{Y}
$$

Then adjust correct class:

$$
\delta_{i, y_i} = \delta_{i, y_i} - 1
$$

This matches the code:

```
delta3 = probs
delta3[range(len(X)), y] -= 1
```

---

### 🔑 Key Insight

Softmax + Cross-Entropy gives a very clean gradient:

$$
\hat{Y} - Y
$$

This avoids explicitly computing derivatives of exponentials and greatly simplifies implementation.

---

### 10. Cost Tracking

At each iteration:

$$
J = \frac{1}{n} (\text{data loss} + \text{regularization loss})
$$

Cost is stored and plotted against iterations.

````python
def softmax_(y):
    if len(y.shape) == 1:
        axis = 0  # colmn wise
    else:
        axis = 1  # row wise
    y_max = np.max(y, axis=axis, keepdims=True)
    y_shift = np.exp(y - y_max)
    y_shift_sum = np.sum(y_shift, axis=axis, keepdims=True)
    return y_shift / y_shift_sum

class SimpleLogisticRegression:
    def __init__(self, input_size=4, hidden_size=3, reg=None):
        self.m = np.random.randn(input_size, hidden_size) / np.sqrt(input_size)
        self.b = np.zeros(hidden_size)
        self.reg = reg
        self.alpha = 1.0

    def predict(self, X):

        z = np.dot(X, self.m) + self.b
        y_pred = softmax_(z)
        return y_pred

    def get_cost(self, X, y):
        probs = self.predict(X)
        correct_logprobs = -np.log(probs[range(len(X)), y])
        data_loss = np.sum(correct_logprobs)
        reg_loss = 0
        if self.reg == "l2":
            reg_loss = 0.5 * self.alpha * np.sum(self.m**2)
        return (1.0 / len(X)) * (data_loss + reg_loss)

    def fit(self, X, y, learning_rate=0.001, iterations=100):

        cost = []

        for i in range(iterations):
            z = np.dot(X, self.m) + self.b
            probs = softmax_(z)

            delta3 = probs
            delta3[range(len(X)), y] -= 1

            dm = np.dot(X.T, delta3)
            if self.reg == "l2":
                dm += self.alpha * self.m

            db = np.sum(delta3, axis=0)

            self.m = self.m - (learning_rate * dm)
            self.b = self.b - (learning_rate * db)

            cost.append(self.get_cost(X, y))
            if (i + 1) % 25 == 0:
                print("Cost at iteration number " + str(i + 1) + " is : ", cost[i])
        plt.title("cost v/s iteration")
        plt.plot(cost)
        plt.xlabel("no. of iterations")
        plt.ylabel("cost")
        plt.show()
````

````python
model = SimpleLogisticRegression(input_size=n_features, hidden_size=n_classes)
````

````python
model.fit(x, y)
````

````python
import numpy as np


class TwoLayerNet:
    def __init__(self, input_size, hidden_size, output_size=1):
        self.params = {}
        # Xavier initialization
        self.params["W1"] = np.random.randn(input_size, hidden_size) / np.sqrt(
            input_size
        )
        self.params["W2"] = np.random.randn(hidden_size, output_size) / np.sqrt(
            hidden_size
        )

        # self.params["W1"] = np.random.randn(input_size, hidden_size)
        self.params["b1"] = np.zeros((1, hidden_size))
        # self.params["W2"] = np.random.randn(hidden_size, output_size)
        self.params["b2"] = np.zeros((1, output_size))

    def forward(self, X):
        W1, b1 = self.params["W1"], self.params["b1"]
        W2, b2 = self.params["W2"], self.params["b2"]
        z1 = np.dot(X, W1) + b1
        a1 = np.maximum(0, z1)  # ReLU activation function
        z2 = np.dot(a1, W2) + b2

        return z2

    def loss(self, X, y_true):
        y_pred = self.forward(X)
        cost = (np.sum(np.square(y_pred - y_true))) / (X.shape[0])

        return cost

    def train(self, X, y, num_epochs, learning_rate=0.0001):
        for epoch in range(num_epochs):
            # Forward propagation
            z1 = np.dot(X, self.params["W1"]) + self.params["b1"]
            a1 = np.maximum(0, z1)  # ReLU activation function
            z2 = np.dot(a1, self.params["W2"]) + self.params["b2"]

            delta3 = z2 - y  # ypred-yactual derivative MSE

            dW2 = np.dot(a1.T, delta3)  # layer 2 derivative
            db2 = np.sum(delta3, axis=0)

            delta2 = np.dot(delta3, self.params["W2"].T) * (
                a1 > 0
            )  # derivative of ReLU

            dW1 = np.dot(X.T, delta2)  # layer 1 derivative
            db1 = np.sum(delta2, axis=0)

            # Update parameters
            self.params["W1"] -= learning_rate * dW1
            self.params["b1"] -= learning_rate * db1
            self.params["W2"] -= learning_rate * dW2
            self.params["b2"] -= learning_rate * db2

            # Print loss for monitoring training progress
            # if epoch % 2 == 0:
            loss = self.loss(X, y)
            print("Epoch {}: loss = {}".format(epoch, loss))
````

**Two layer Neural Network (Classification)**

````python
import numpy as np


class TwoLayerNet:
    def __init__(self, input_size, hidden_size, output_size):
        self.params = {}
        # Xavier initialization
        self.params["W1"] = np.random.randn(input_size, hidden_size) / np.sqrt(
            input_size
        )
        self.params["W2"] = np.random.randn(hidden_size, output_size) / np.sqrt(
            hidden_size
        )

        # self.params["W1"] = np.random.randn(input_size, hidden_size)
        self.params["b1"] = np.zeros((1, hidden_size))
        # self.params["W2"] = np.random.randn(hidden_size, output_size)
        self.params["b2"] = np.zeros((1, output_size))

    def forward(self, X):
        W1, b1 = self.params["W1"], self.params["b1"]
        W2, b2 = self.params["W2"], self.params["b2"]
        z1 = np.dot(X, W1) + b1
        a1 = np.maximum(
            z1, 0.01 * z1
        )  # ReLU activation function np.maximum(0,z1) leaky relu np.maximum(alpha*arr, arr)
        z2 = np.dot(a1, W2) + b2

        probs = softmax_(z2)  # softmax calculation

        return probs

    def loss(self, X, y):
        probs = self.forward(X)
        correct_logprobs = -np.log(probs[range(len(X)), y])
        data_loss = np.sum(correct_logprobs)
        return (1.0 / len(X)) * data_loss

    def train(self, X, y, num_epochs, learning_rate=0.0001):
        for epoch in range(num_epochs):
            # Forward propagation
            z1 = np.dot(X, self.params["W1"]) + self.params["b1"]
            a1 = np.maximum(0, z1)  # ReLU activation function
            z2 = np.dot(a1, self.params["W2"]) + self.params["b2"]

            probs = softmax_(z2)

            delta3 = probs

            # back propagation start

            delta3[range(len(X)), y] -= 1  # ypred-yactual derivative of softmax loss

            dW2 = np.dot(a1.T, delta3)  # layer 2 derivative
            db2 = np.sum(delta3, axis=0)

            delta2 = np.dot(delta3, self.params["W2"].T) * (
                np.where(a1 > 0, 1, 0.01)
            )  # derivative of ReLU (a1>0) leaky relu np.where(arr > 0, 1, 0.01)

            dW1 = np.dot(X.T, delta2)  # layer 1 derivative
            db1 = np.sum(delta2, axis=0)

            # Update parameters
            self.params["W1"] -= learning_rate * dW1
            self.params["b1"] -= learning_rate * db1
            self.params["W2"] -= learning_rate * dW2
            self.params["b2"] -= learning_rate * db2

            # Print loss for monitoring training progress
            if epoch % 2 == 0:
                loss = self.loss(X, y)
                print("Epoch {}: loss = {}".format(epoch, loss))
````

````python
net.train(x, y, num_epochs=15)
````

**KNN**

````python
from collections import Counter
import numpy as np


class KNN:
    def __init__(self, k=3, distance="euclidean"):
        self.k = k
        self.distance = distance

    def fit(self, X_train, y_train):
        self.X_train = X_train
        self.y_train = y_train

    def predict(self, X_test):
        y_pred = []
        for x in X_test:
            # Compute distances between the test point and all training points
            if self.distance == "euclidean":
                distances = np.linalg.norm(self.X_train - x, axis=1)
            elif self.distance == "manhattan":
                distances = np.sum(np.abs(self.X_train - x), axis=1)

            # Select the k nearest neighbors
            nearest_indices = np.argsort(distances)[: self.k]
            nearest_labels = self.y_train[nearest_indices]

            # Assign the class label that appears most frequently among the k nearest neighbors
            label = Counter(nearest_labels).most_common(1)[0][0]
            y_pred.append(label)

        return np.array(y_pred)
````

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load the iris dataset
iris = load_iris()

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

# Create a KNN classifier with k=5 and euclidean distance
knn = KNN(k=5, distance="euclidean")

# Train the classifier on the training data
knn.fit(X_train, y_train)

# Make predictions on the test data
y_pred = knn.predict(X_test)

# Compute the accuracy of the classifier
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
````

**Kmeans**

````python
import numpy as np


class KMeans:
    def __init__(self, k=3, max_iters=100, tol=1e-4):
        self.k = k
        self.max_iters = max_iters
        self.tol = tol

    def fit(self, X):
        # Initialize centroids randomly
        self.centroids = X[
            np.random.choice(X.shape[0], self.k, replace=False)
        ]  # k*feature

        # Iterate until convergence or maximum number of iterations is reached
        for i in range(self.max_iters):
            # Assign each data point to the closest centroid
            distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)  # n*k
            cluster_assignments = np.argmin(distances, axis=1)  #

            # Update the centroids based on the new cluster assignments
            new_centroids = np.array(
                [
                    np.mean(X[np.where(cluster_assignments == j)], axis=0)
                    for j in range(self.k)
                ]
            )

            # Check for convergence
            if np.linalg.norm(new_centroids - self.centroids) < self.tol:
                break

            self.centroids = new_centroids

    def predict(self, X):
        # Assign each data point to the closest centroid
        distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
        cluster_assignments = np.argmin(distances, axis=1)

        return cluster_assignments
````

````python
x1 = np.random.randn(5, 2) + 5
x2 = np.random.randn(5, 2) - 5
X = np.concatenate([x1, x2], axis=0)
print(X.shape)
````

````python
# Initialize the KMeans object with k=3
kmeans = KMeans(k=2)

# Fit the k-means model to the dataset
kmeans.fit(X)

# Get the cluster assignments for the input dataset
cluster_assignments = kmeans.predict(X)

# Print the cluster assignments
print(cluster_assignments)

# Print the learned centroids
print(kmeans.centroids)
````
